```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2")
model = AutoModelForCausalLM.from_pretrained("openai-community/gpt2")
```

```python
# Setup
import torch
tokenizer.pad_token = tokenizer.eos_token
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
print(f"Model loaded on {device}")
```

### 2. Prompt Fitering

```python
def create_prompt(user_question):
    prompt = f"""You are a Python Coding Assistant. You can ONLY answer the questions about the Python Programing.
    Question : {user_question}

Instructions : 
1. If the question is all about the phython then provide as much as possible helpful answer to the user.
2. but if the question is not related to the python then just reply in the same exact manner : "I can only answer about the Python Programming related questions."

Answer:
"""
    return prompt
```

### 3. Getting the response

```python
def generate_response(user_question):
    prompt = create_prompt(user_question)

    # Tokenize and generate
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    outputs = model.generate(**inputs, max_new_tokens=150, temperature=0.7, do_sample=True)
    
    # Decode and extract answer
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response.split("Answer:")[-1].strip()
```

### 4.Test it

```python
q1 = "How do I create a list in Python?"
print(f"Q: {q1}")
print(f"A: {generate_response(q1)}\n")
```

```python
# Test 4: General knowledge (should reject)
print("="*70)
print("TEST 4: General Knowledge Question")
print("="*70)
question4 = "Who won the World Cup?"
print(f"Question: {question4}")
response4 = generate_response(question4)
print(f"Answer: {response4}\n")
```
